```python
df = pd.read_csv("superstore_sales.csv")

print("Dataset Loaded Successfully")
print("Shape:", df.shape)
display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (missing_values / len(df)) * 100
})

display(missing_report[missing_report["Missing Values"] > 0])
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())

print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())
```

```python
print(df["Row ID"].value_counts().head(10))

display(df[df["Row ID"] == "Yes"].head())
```

```python
df_clean = df.dropna(subset=["Order Date"])

df_clean = df_clean.copy()

df_clean["Row ID"] = pd.to_numeric(df_clean["Row ID"])

print("Original Dataset:", df.shape)
print("Cleaned Dataset:", df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nDuplicate Rows:", df_clean.duplicated().sum())

print("\nDuplicate Row IDs:", df_clean["Row ID"].duplicated().sum())
```

```python
df_clean["Order Date"] = pd.to_datetime(df_clean["Order Date"], format="mixed", errors="coerce")
df_clean["Ship Date"] = pd.to_datetime(df_clean["Ship Date"], format="mixed", errors="coerce")

df_clean["Year"] = df_clean["Order Date"].dt.year
df_clean["Month"] = df_clean["Order Date"].dt.month
df_clean["Year-Month"] = df_clean["Order Date"].dt.to_period("M").astype(str)

print("Date Conversion Completed")

print("\nData Types:")
print(df_clean[["Order Date", "Ship Date"]].dtypes)

print("\nDate Range:")
print("Start:", df_clean["Order Date"].min())
print("End:", df_clean["Order Date"].max())

display(df_clean[["Order Date", "Ship Date", "Year", "Month", "Year-Month"]].head())
```

```python
monthly_sales = df_clean.groupby("Year-Month")["Sales"].sum().reset_index()

monthly_sales = monthly_sales.sort_values("Year-Month")

display(monthly_sales)

plt.figure(figsize=(14, 6))
plt.plot(monthly_sales["Year-Month"], monthly_sales["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
highest_month = monthly_sales.loc[monthly_sales["Sales"].idxmax()]
lowest_month = monthly_sales.loc[monthly_sales["Sales"].idxmin()]

print("Highest Sales Month:")
print(highest_month)

print("\nLowest Sales Month:")
print(lowest_month)
```

```python
category_sales = df_clean.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

category_sales = category_sales.sort_values(
    by="Total_Sales", ascending=False
)

display(category_sales)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_sales,
    x="Category",
    y="Total_Sales"
)

plt.title("Category-wise Revenue Analysis")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
region_analysis = df_clean.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

region_analysis = region_analysis.sort_values(
    by="Total_Sales", ascending=False
)

display(region_analysis)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=region_analysis,
    x="Region",
    y="Total_Sales"
)

plt.title("Region-wise Sales Analysis")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=region_analysis,
    x="Region",
    y="Total_Profit"
)

plt.title("Region-wise Profit Analysis")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
segment_analysis = df_clean.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Total_Orders=("Order ID", "nunique")
).reset_index()

display(segment_analysis)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=segment_analysis,
    x="Segment",
    y="Total_Sales"
)

plt.title("Customer Segment-wise Sales Analysis")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=segment_analysis,
    x="Segment",
    y="Total_Profit"
)

plt.title("Customer Segment-wise Profit Analysis")
plt.xlabel("Customer Segment")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
discount_analysis = df_clean.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

display(discount_analysis)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=discount_analysis,
    x="Discount",
    y="Average_Profit"
)

plt.title("Discount vs Average Profit Analysis")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.tight_layout()
plt.show()
```

```python
customer_analysis = df_clean.groupby("Customer Name").agg(
    Total_Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Total_Profit=("Profit", "sum")
).reset_index()

customer_analysis = customer_analysis.sort_values(
    by="Total_Orders", ascending=False
)

display(customer_analysis.head(10))
```

```python
top_customers = customer_analysis.head(10)

plt.figure(figsize=(12, 6))
sns.barplot(data=top_customers, x="Total_Orders", y="Customer Name")
plt.title("Top 10 Customers by Number of Orders")
plt.xlabel("Total Orders")
plt.ylabel("Customer Name")
plt.tight_layout()
plt.show()
```
